```python
import re
import pandas as pd

# Function to normalize item names
def normalize_item_name(name):
    if pd.isna(name):
        return ""

    name = str(name).lower()

    # Remove prices like 'Rs. 40'
    name = re.sub(r'rs\.\s*\d+(\.\d+)?', '', name)

    # Remove units + quantities: '21g', '4 ml', '50 pcs', etc.
    name = re.sub(r'\d+\s*(g|gm|kg|ml|pcs|ltr|litre)', '', name)

    # Remove count indicators: x1, x2 ... x99
    name = re.sub(r'\bx\d{1,2}\b', '', name)

    # Remove packaging-related words
    name = re.sub(r'\b(pack|packs|box|boxes|sachet|carton|bottle|bar|wrapper|dozen)\b', '', name)

    # Remove size adjectives
    name = re.sub(r'\b(super size|small size|regular size|large size|big size|mini size|medium size|m|s|l|xl|xxl)\b', '', name)

    # Remove content inside parentheses
    name = re.sub(r'\([^)]*\)', '', name)

    # Remove standalone numbers
    name = re.sub(r'\b\d+\b', '', name)

    # Remove extra "x" if isolated
    name = re.sub(r'\bx\b', '', name)

    # Normalize whitespace
    name = re.sub(r'\s+', ' ', name).strip()

    return name

# Apply the function
DATASET_2['distinct_item'] = DATASET_2['item_name'].apply(normalize_item_name)

# Double-check that distinct_item has values
if 'distinct_item' in DATASET_2.columns:
    item_counts = DATASET_2['distinct_item'].value_counts().reset_index()
    item_counts.columns = ['distinct_item', 'count']

    # Merge back into original DataFrame
    DATASET_2 = pd.merge(DATASET_2, item_counts, on='distinct_item', how='left')

    # Show result
    print(DATASET_2[['distinct_item', 'count']].head(10))
else:
    print("Column 'distinct_item' not created properly.")
```
